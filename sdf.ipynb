```python
x, y, z = var('x y z')

c = var('c0 c1 c2 c3 c4 c5 c6 c7')

f = (
    c[0] * (1-x) * (1-y) * (1-z) +
    c[1] * x     * (1-y) * (1-z) +
    c[2] * (1-x) * y     * (1-z) +
    c[3] * x     * y     * (1-z) +
    c[4] * (1-x) * (1-y) * z +
    c[5] * x     * (1-y) * z +
    c[6] * (1-x) * y     * z +
    c[7] * x     * y     * z
)

def make_surface(values):
    return f.subs({c[i]: values[i] for i in range(8)})
```

```python
# Corner values, 8bit, red sphere with blue marked as removal,taken from shader debug array
u = [-0.000330, 0.001270, -0.000730, 0.001270, -0.000330, 0.001070, -0.000930, 0.001070]
u_sd = make_surface(u)
```

```python
# 8bit values, blue shphere only, red moved away, values from debug array in shader
v = [0.000470, 0.000070, 0.002270, 0.001870, 0.000270, -0.000130, 0.002270, 0.001870]
v_sd = make_surface(v)
```

```python
import random as pyrandom
def sample_surface(f, u, v, N=100):
    points = []

    for i in range(N):
        X = pyrandom.random()
        Y = pyrandom.random()
        Z = pyrandom.random()

        # Try the z direction
        f0 = f(x=X, y=Y, z=0)
        f1 = f(x=X, y=Y, z=1)

        if f0 * f1 <= 0 and f0 != f1:
            Zs = -f0 / (f1 - f0)

            if u(x=X, y=Y, z=Zs) < v(x=X, y=Y, z=Zs):
                points.append((X, Y, Zs))

        # Try x direction
        f0 = f(x=0, y=Y, z=Z)
        f1 = f(x=1, y=Y, z=Z)

        if f0 * f1 <= 0 and f0 != f1:
            Xs = -f0 / (f1 - f0)

            if u(x=Xs, y=Y, z=Z) < v(x=Xs, y=Y, z=Z):
                points.append((Xs, Y, Z))

        # Try y direction
        f0 = f(x=X, y=0, z=Z)
        f1 = f(x=X, y=1, z=Z)

        if f0 * f1 <= 0 and f0 != f1:
            Ys = -f0 / (f1 - f0)

            if u(x=X, y=Ys, z=Z) < v(x=X, y=Ys, z=Z):
                points.append((X, Ys, Z))

    return points

points_u_lt_v = sample_surface(muv_sd, u_sd, v_sd, 2000)
points_v_lt_u = sample_surface(muv_sd, v_sd, u_sd, 2000)

axes = (
    line3d([(0,0,0), (1,0,0)], color='red') +
    line3d([(0,0,0), (0,1,0)], color='green') +
    line3d([(0,0,0), (0,0,1)], color='blue')
)

(point3d(points_u_lt_v, size=2, color='red') + point3d(points_v_lt_u, size=2, color='blue') + axes).show(
    aspect_ratio=(1,1,1)
)
```

```python
def cube_from_corner(corner_pos, side,color,opacity): # corner is low coordinate corner
    return cube((corner_pos[0] + 0.5 * side, corner_pos[1] + 0.5 * side, corner_pos[2] + 0.5 * side), side, color=color, opacity=opacity)

    
def voxel_id_to_world(x,y,z): # position of lower corner
    return (inter_sample_distance * 0.5 + inter_sample_distance * x,
     inter_sample_distance * 0.5 + inter_sample_distance * y,
     inter_sample_distance * 0.5 + inter_sample_distance * z)
    
def draw_voxel_material_and_surfaces(origin_voxel, radius):
    origin_voxel_pos = voxel_id_to_world(*origin_voxel)
    region = [
        (x,origin_voxel_pos[0], origin_voxel_pos[0] + radius * inter_sample_distance),
        (y,origin_voxel_pos[1], origin_voxel_pos[1] + radius * inter_sample_distance),
        (z,origin_voxel_pos[2], origin_voxel_pos[2] + radius * inter_sample_distance)        
    ]
    graphics_3d = implicit_plot3d(
        red_minus_blue_sd,
        *region,
        color='red', opacity=1, plot_points=40
    )
    graphics_3d += implicit_plot3d(
        blue_sd,
        *region,
        color='blue', opacity=1, plot_points=40
    )
    for i in range(0, radius):
        for j in range(0, radius):
            for k in range(0,radius):
                voxel_id = (origin_voxel[0]+i, origin_voxel[1]+j, origin_voxel[2]+k)
                mat_info = voxels[voxel_id[2],voxel_id[1],voxel_id[0]]
                if mat_info >= 64 and mat_info <= 95: # C12 config
                    graphics_3d += cube_from_corner(voxel_id_to_world(*voxel_id), inter_sample_distance, color='yellow', opacity=0.6)
                #if mat_info == 3: 
                    #graphics_3d += cube_from_corner(voxel_id_to_world(*voxel_id), inter_sample_distance, color='violet', opacity=0.3)
                #if mat_info == 2: 
                    #graphics_3d += cube_from_corner(voxel_id_to_world(*voxel_id), inter_sample_distance, color='pink', opacity=0.3)

    return graphics_3d
                
```
